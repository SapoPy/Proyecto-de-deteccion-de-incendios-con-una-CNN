```python
import os
import sys
import time

import numpy as np
import torch
from torch import nn
import torchvision
from matplotlib import pyplot as plt

from utils import *
from Modelos.fireyecnn import *

from dataset import *

import copy
```

```python
np.random.seed(13117532)

fireeye_train, fireeye_val, fireeye_test = create_datasets(
    DIR_TRAIN,
    DIR_TEST,
    transform=transform_fireeye
)



indices_train = np.random.permutation(len(fireeye_train))
indices_val = np.random.permutation(len(fireeye_val))

small_indices_train = indices_train[:6000]
small_indices_val = indices_val[:600]

fireeye_train_small = copy.deepcopy(fireeye_train)
fireeye_val_small =  copy.deepcopy(fireeye_val)


fireeye_train_small.samples = [
    fireeye_train_small.samples[i]
    for i in small_indices_train
]

fireeye_val_small.samples = [
    fireeye_val_small.samples[i]
    for i in small_indices_val
]

lr = 5e-4
dropout_p = 0.5
batch_size = 16
criterion = nn.CrossEntropyLoss()
use_gpu = False

epochs = 15

model = CNNModel(dropout_p)
```

```python
curves = train_model(
    model,
    fireeye_train,
    fireeye_val,
    epochs,
    criterion,
    batch_size,
    lr,
    use_gpu=use_gpu,
    data_augmentation=False,
)

show_curves(curves)
```
