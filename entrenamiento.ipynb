```python
import os
import sys
import time

import numpy as np
import torch
from torch import nn
import torchvision
from matplotlib import pyplot as plt

from utils import *
from Modelos.fireyecnn import *

from dataset import *

import copy
```

```python

fireeye_train = FireEyeDataset(DIR_TRAIN, transform=transform_fireeye)
fireeye_test = FireEyeDataset(DIR_TEST,transform=transform_fireeye)

np.random.seed(1234567)
indices_train = np.random.permutation(len(fireeye_train))
indices_test = np.random.permutation(len(fireeye_test))

small_indices_train = indices_train[:500]
small_indices_test = indices_test[:50]

fireeye_train_small = copy.deepcopy(fireeye_train)
fireeye_test_small =  copy.deepcopy(fireeye_test)


fireeye_train_small.samples = [
    fireeye_train_small.samples[i]
    for i in small_indices_train
]

fireeye_test_small.samples = [
    fireeye_test_small.samples[i]
    for i in small_indices_test
]

lr = 5e-4
dropout_p = 0.5
batch_size = 4
criterion = nn.CrossEntropyLoss()

epochs = 30

model = CNNModel(dropout_p)

curves = train_model(
    model,
    fireeye_train_small,
    fireeye_test_small,
    epochs,
    criterion,
    batch_size,
    lr,
    use_gpu=False,
    data_augmentation=False,
)

show_curves(curves)
```
