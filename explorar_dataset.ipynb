```python
import numpy as np
import matplotlib.pyplot as plt
import dataset as fe
from PIL import Image

fireeye_train = fe.FireEyeDataset(fe.DIR_TRAIN)
fireeye_test = fe.FireEyeDataset(fe.DIR_TEST)
```

```python
print("Total de imagenes en el dataset:", len(fireeye_train) + len(fireeye_test))
print("Total imágenes de train:", len(fireeye_train))
print("Total imágenes de test:", len(fireeye_test))
```

```python
print(fireeye_train.class_to_idx)
```

```python
n_per_class = 5
samples_per_class = {i: [] for i in range(len(fireeye_train.classes))}

for img_path, label in fireeye_train.samples:
    if len(samples_per_class[label]) < n_per_class:
        samples_per_class[label].append(img_path)

# Plot
plt.figure(figsize=(10, 6))
plot_idx = 1

for label, paths in samples_per_class.items():
    for img_path in paths:
        img = Image.open(img_path).convert("RGB")
        
        plt.subplot(len(samples_per_class), n_per_class, plot_idx)
        plt.imshow(img)
        plt.title(fireeye_train.classes[label])
        plt.axis('off')
        
        plot_idx += 1

plt.tight_layout()
plt.show()
```

```python
class_counts = {i: 0 for i in range(len(fireeye_train.classes))}

for _, label in fireeye_train.samples:
    class_counts[label] += 1

print(class_counts)
```

```python
plt.figure(figsize=(12, 5))

labels = [fireeye_train.classes[i] for i in class_counts.keys()]
values = list(class_counts.values())

plt.bar(labels, values)

plt.title("Distribución de imágenes por clase")
plt.xlabel("Clases")
plt.ylabel("Cantidad de imágenes")

plt.show()
```
